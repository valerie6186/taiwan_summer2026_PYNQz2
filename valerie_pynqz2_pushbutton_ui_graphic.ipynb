PYNQ-Z2 Pushbutton UI Interface</div>

<div style="background-color: #f3e5ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
This notebook will creates four UI graphics that correspond to the four physical pushbuttons on the PYNQ-Z2 board.</div>

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 1: Import the Required Libraries </h3> 
<ul> <li><b>BaseOverlay</b> provides access to the built-in hardware components on the PYNQ-Z2 board</li> 
<li><b>ipywidgets</b> is used to create the button graphics inside Jupyter Notebook</li> 
<li><b>display</b> is used to show the UI graphics in the notebook</li> <li><b>time</b> is used to pause briefly between button readings</li> </ul> </div>

```python
from pynq.overlays.base import BaseOverlay 
import ipywidgets as widgets 
from IPython.display import display 
import time
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 2: Load the PYNQ-Z2 Base Overlay </h3> 
<ul> <li>an overlay is a hardware design that is loaded onto the FPGA portion of the PYNQ-Z2 board</li> 
<li>the base overlay provides Python access to the four physical pushbuttons</li> 
<li>this notebook must be running directly on the PYNQ-Z2 board</li> </ul> </div>

```python
# load the PYNQ-Z2 base overlay
base = BaseOverlay('base.bit')
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 3: Create Objects for the Four Physical Pushbuttons </h3> 
<ul> <li><b>base.buttons</b> contains the four physical pushbuttons provided by the base overlay</li> 
<li>each pushbutton is accessed using an index from 0 through 3</li> <li>the <b>read()</b> method will later check whether each button is pressed or released</li> </ul> </div>

```python
# create an object for each physical pushbutton 
btn0 = base.buttons[0] 
btn1 = base.buttons[1] 
btn2 = base.buttons[2] 
btn3 = base.buttons[3]
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 4: Create the Four UI Button Graphics </h3> 
<ul> <li>each <b>ToggleButton</b> graphic represents one physical pushbutton on the PYNQ-Z2 board</li> 
<li><b>disabled=True</b> prevents the user from changing the graphic by clicking it with the mouse</li> 
<li>the UI graphics will only respond to the physical pushbuttons</li> <li>the initial gray appearance means the corresponding physical button is not pressed</li> </ul> </div>

```python
# create the BTN0 UI graphic 
btn0_ui = widgets.ToggleButton( 
    value=False, 
    description='BTN0: not pressed', 
    disabled=True, 
    button_style='', 
    icon='circle-o', 
    layout=widgets.Layout( 
        width='185px', 
        height='70px', 
        margin='0 8px' 
    ) 
) 
# create the BTN1 UI graphic 
btn1_ui = widgets.ToggleButton( 
    value=False, 
    description='BTN1: not pressed', 
    disabled=True, 
    button_style='', 
    icon='circle-o', 
    layout=widgets.Layout( 
        width='185px', 
        height='70px', 
        margin='0 8px' 
    ) 
) 
# create the BTN2 UI graphic 
btn2_ui = widgets.ToggleButton( 
    value=False, 
    description='BTN2: not pressed', 
    disabled=True, 
    button_style='', 
    icon='circle-o', 
    layout=widgets.Layout( 
        width='185px', 
        height='70px', 
        margin='0 8px' 
    ) 
) 
# create the BTN3 UI graphic 
btn3_ui = widgets.ToggleButton( 
    value=False, 
    description='BTN3: not pressed', 
    disabled=True, button_style='', 
    icon='circle-o', layout=widgets.Layout( 
        width='185px', 
        height='70px', 
        margin='0 8px' 
    ) 
)
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 5: Arrange the Four UI Graphics Horizontally </h3> 
<ul> <li>the four graphics are placed in one horizontal row</li> <li>the UI graphics are ordered from BTN0 through BTN3</li> 
<li>the horizontal layout makes it easier to match each UI graphic with its corresponding physical board button</li> </ul> </div>

```python
# arrange the four UI graphics in one horizontal row 
button_row = widgets.HBox( 
    children=[ 
        btn3_ui,
        btn2_ui,
        btn1_ui,
        btn0_ui
    ], 
    layout=widgets.Layout( 
        display='flex', 
        flex_flow='row', 
        justify_content='center', 
        align_items='center', 
        width='100%' 
    ) 
) 

display(button_row)
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 6: Match Each Physical Button with Its UI Graphic </h3> 
<ul> <li>each dictionary stores one physical button, its matching UI graphic, and its display label</li> 
<li>the dictionaries are stored together in one list</li> 
<li>this allows all four buttons to be checked using one loop instead of repeating the same code four times</li> </ul> </div>

```python
# connect each physical button to its matching UI graphic 
button_information = [
    { 'button': btn0,
      'ui': btn0_ui, 
      'label': 'BTN0' 
    }, 
    { 'button': btn1, 
      'ui': btn1_ui, 
      'label': 'BTN1' 
    }, 
    { 'button': btn2, 
      'ui': btn2_ui,  
      'label': 'BTN2' 
    }, 
    { 'button': btn3, 
      'ui': btn3_ui,  
      'label': 'BTN3' 
    } 
]
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;"> 
<h3 style="font-size: 20px; font-weight: bold;"> Step 7: Monitor the Four Physical Pushbuttons </h3> 
<ul> <li>the program continuously reads the four physical pushbuttons</li> 
<li><b>button.read()</b> returns 1 when a button is pressed and 0 when it is released</li> 
<li>when a button is pressed, its corresponding UI graphic becomes green</li> 
<li>the graphic changes from a circle icon to a checkmark and displays <b>PRESSED</b></li> 
<li>when the physical button is released, the UI graphic returns to gray</li> 
<li>click Jupyter's Stop button or interrupt the kernel to stop the detection loop</li> </ul> </div>

```python
try:
    print("Starting four-button detection...")
    print("Press Jupyter's STOP button or interrupt the kernel to stop")

    while True:
        # check each physical pushbutton
        for item in button_information:
            pressed = item['button'].read() == 1

            if pressed:
                item['ui'].value = True
                item['ui'].description = item['label'] + ': PRESSED'
                item['ui'].button_style = 'success'
                item['ui'].icon = 'check-circle'

            else:
                item['ui'].value = False
                item['ui'].description = item['label'] + ': not pressed'
                item['ui'].button_style = ''
                item['ui'].icon = 'circle-o'

        # pause briefly to reduce unnecessary CPU usage
        time.sleep(0.05)

except KeyboardInterrupt:
    # return all UI graphics to their original appearance
    for item in button_information:
        item['ui'].value = False
        item['ui'].description = item['label'] + ': not pressed'
        item['ui'].button_style = ''
        item['ui'].icon = 'circle-o'

    print("\nbutton detection successfully stopped")
```

<li>releasing a button should return its UI graphic to gray</li> </ul> </div>
