    //  1. 4-Bit Adder Design Module
        module adder_4bit (
            input  [3:0] a,
            input  [3:0] b,
            input        cin,
            output [3:0] sum,
            output       cout
    );
        // use behavioral-level description to perform a 5-bit addition
        assign {cout, sum} = a + b + cin;
    endmodule


    // 2. Testbench Module
    module testbench;

        // Declare input signals as register types
        reg [3:0] a;
        reg [3:0] b;
        reg       cin;

        // Declare output signals as wire types
        wire [3:0] sum;
        wire       cout;

        // Instantiate the Unit Under Test (UUT)
        adder_4bit uut (
            .a(a),
            .b(b),
            .cin(cin),
            .sum(sum),
            .cout(cout)
        );

        // 3. Begin applying test inputs
        initial begin

            // display the output table headings
            $display("Time\t A \t B \t Cin \t Sum \t Cout");
            $display("----------------------------------------------");

            // Test case 1: 0 + 0 = 0
            a = 4'b0000;
            b = 4'b0000;
            cin = 0;

            #10; // wait for 10 simulation time units

            $display(
                "%0d\t %b \t %b \t %b \t %b \t %b",
                $time, a, b, cin, sum, cout
            );

            // test case 2: 5 + 3 = 8
            a = 4'd5;
            b = 4'd3;
            cin = 0;

            #10;

            $display(
                "%0d\t %b \t %b \t %b \t %b \t %b",
                $time, a, b, cin, sum, cout
            );

            // test case 3: 15 + 1 = 16, producing a carry-out
            a = 4'd15;
            b = 4'd1;
            cin = 0;

            #10;

            $display(
                "%0d\t %b \t %b \t %b \t %b \t %b",
                $time, a, b, cin, sum, cout
            );

            // test case 4: 7 + 8 + 1 = 16, producing a carry-out
            a = 4'd7;
            b = 4'd8;
            cin = 1;

            #10;

            $display(
                "%0d\t %b \t %b \t %b \t %b \t %b",
                $time, a, b, cin, sum, cout
            );

            $finish; // End the simulation
        end
    endmodule
