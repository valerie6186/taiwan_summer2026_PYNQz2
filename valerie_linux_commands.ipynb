Linux Commands</div>

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; "> 
<h3 style="font-size: 20px; font-weight: bold;">Navigation</h3>
<br>
<strong>pwd</strong> <ul> <li>shows the full path of the directory you are currently in</li> </ul> <strong>ls</strong> <ul> <li>lists files and folders in the current directory</li> </ul> <strong>ls -l</strong> <ul> <li>lists files and folders with details such as permissions, owner, size, and date</li> </ul> <strong>ls -a</strong> <ul> <li>shows all files, including hidden files</li> </ul> <strong>ls -la</strong> <ul> <li>shows all files, including hidden files, with detailed information</li> </ul> <strong>cd folder_name</strong> <ul> <li>moves into a folder</li> </ul> <strong>cd ..</strong> <ul> <li>moves up one directory level</li> </ul> <strong>cd ~</strong> <ul> <li>moves to your home directory</li> </ul> <strong>cd /</strong> <ul> <li>moves to the root directory</li> </ul> <strong>cd -</strong> <ul> <li>returns to the previous directory</li> </ul> </div>

<div style="background-color: #FFDFBE; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">

<h3 style="font-size: 20px; font-weight: bold;">Creating Files and Folders</h3>
<br>
<strong>mkdir folder_name</strong>
<ul>
<li>creates a new folder</li>
</ul>

<strong>mkdir -p folder1/folder2</strong>
<ul>
<li>creates multiple nested folders at once</li>
</ul>

<strong>touch file.txt</strong>
<ul>
<li>creates a new empty file</li>
</ul>

<strong>nano file.txt</strong>
<ul>
<li>opens or creates a file using the Nano text editor</li>
</ul>

<strong>vim file.txt</strong>
<ul>
<li>opens or creates a file using the Vim text editor</li>
</ul>

</div>

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Viewing Files</h3> 
<br>
<strong>cat file.txt</strong> <ul> <li>displays the entire contents of a file</li> </ul> <strong>less file.txt</strong> <ul> <li>displays a file one page at a time</li> </ul> <strong>head file.txt</strong> <ul> <li>displays the first 10 lines of a file</li> </ul> <strong>head -n 5 file.txt</strong> <ul> <li>displays the first five lines of a file</li> </ul> <strong>tail file.txt</strong> <ul> <li>displays the last 10 lines of a file</li> </ul> <strong>tail -f file.txt</strong> <ul> <li>continuously displays new lines added to a file, which is useful for logs</li> </ul> </div>

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Copying, Moving, and Renaming</h3> 
<br>
<strong>cp file1.txt file2.txt</strong> <ul> <li>copies file1.txt and names the copy file2.txt</li> </ul> <strong>cp file.txt folder_name/</strong> <ul> <li>copies a file into another folder</li> </ul> <strong>cp -r folder1 folder2</strong> <ul> <li>copies an entire folder and everything inside it</li> </ul> <strong>mv file.txt folder_name/</strong> <ul> <li>moves a file into another folder</li> </ul> <strong>mv old_name.txt new_name.txt</strong> <ul> <li>renames a file</li> </ul> <strong>mv old_folder new_folder</strong> <ul> <li>renames a folder</li> </ul> </div>

<div style="background-color: #FFDFBE; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Deleting Files and Folders</h3>
<br>
<strong>rm file.txt</strong>
<ul>
<li>permanently deletes a file</li>
</ul>

<strong>rm -i file.txt</strong>
<ul>
<li>asks for confirmation before deleting a file</li>
</ul>

<strong>rm -r folder_name</strong>
<ul>
<li>permanently deletes a folder and everything inside it</li>
</ul>

<strong>rm -rf folder_name</strong>
<ul>
<li>forcefully deletes a folder and everything inside it without confirmation</li>
<li>use this command carefully because deleted files normally cannot be recovered</li>
</ul>

<strong>rmdir folder_name</strong>
<ul>
<li>deletes an empty folder</li>
</ul>

</div>

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Searching</h3> 
<br>    
<strong>find . -name "file.txt"</strong> <ul> <li>searches for a file starting from the current directory</li> </ul> <strong>find . -name "*.py"</strong> <ul> <li>finds all Python files in the current directory and its subfolders</li> </ul> <strong>grep "word" file.txt</strong> <ul> <li>searches for a word or pattern inside a file</li> </ul> <strong>grep -i "word" file.txt</strong> <ul> <li>searches without considering uppercase or lowercase letters</li> </ul> <strong>grep -r "word" folder_name</strong> <ul> <li>searches for text inside every file in a folder</li> </ul> <strong>which python</strong> <ul> <li>shows the location of a command or program</li> </ul> <strong>whereis python</strong> <ul> <li>shows possible locations for a program and its documentation</li> </ul> </div>

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; "> 
<h3 style="font-size: 20px; font-weight: bold;">File and Directory Information</h3> 
<br>
<strong>file filename</strong> <ul> <li>identifies the type of a file</li> </ul> <strong>stat filename</strong> <ul> <li>shows detailed information about a file</li> </ul> <strong>du -h filename</strong> <ul> <li>shows the amount of storage used by a file</li> </ul> <strong>du -sh folder_name</strong> <ul> <li>shows the total storage used by a folder</li> </ul> <strong>df -h</strong> <ul> <li>shows available and used disk space</li> </ul> <strong>wc file.txt</strong> <ul> <li>counts the lines, words, and characters in a file</li> </ul> <strong>wc -l file.txt</strong> <ul> <li>counts only the number of lines in a file</li> </ul> </div>

<div style="background-color: #FFDFBE; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Permissions</h3> 
<br>
<strong>ls -l</strong> <ul> <li>displays file permissions</li> </ul> <strong>chmod +x script.sh</strong> <ul> <li>makes a script executable</li> </ul> <strong>chmod 755 script.sh</strong> <ul> <li>gives the owner full permissions and gives other users read and execute permissions</li> </ul> <strong>chmod 644 file.txt</strong> <ul> <li>gives the owner read and write permissions and gives other users read permission</li> </ul> <strong>sudo command</strong> <ul> <li>runs a command with administrator privileges</li> <li>only use sudo when necessary because it gives the command more control over the system</li> </ul> </div>

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Running Programs</h3> 
<br>    
<strong>./program</strong> <ul> <li>runs an executable program in the current directory</li> </ul> <strong>bash script.sh</strong> <ul> <li>runs a Bash script</li> </ul> <strong>python file.py</strong> <ul> <li>runs a Python file using the Python command</li> </ul> <strong>python3 file.py</strong> <ul> <li>runs a Python file using Python 3</li> </ul> <strong>gcc program.c -o program</strong> <ul> <li>compiles a C program and creates an executable named program</li> </ul> <strong>make</strong> <ul> <li>builds a project using instructions inside a Makefile</li> </ul> </div>

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; "> 
<h3 style="font-size: 20px; font-weight: bold;">Processes and System Information</h3> 
<br>    
<strong>clear</strong> <ul> <li>clears the terminal screen</li> </ul> <strong>history</strong> <ul> <li>shows previously used commands</li> </ul> <strong>history | grep "git"</strong> <ul> <li>searches your command history for commands containing the word git</li> </ul> <strong>whoami</strong> <ul> <li>displays the current username</li> </ul> <strong>hostname</strong> <ul> <li>displays the computer or device name</li> </ul> <strong>uname -a</strong> <ul> <li>displays Linux kernel and system information</li> </ul> <strong>date</strong> <ul> <li>displays the current date and time</li> </ul> <strong>uptime</strong> <ul> <li>shows how long the system has been running</li> </ul> <strong>top</strong> <ul> <li>displays currently running processes and system usage</li> </ul> <strong>htop</strong> <ul> <li>displays running processes in a more user-friendly format if installed</li> </ul> <strong>ps</strong> <ul> <li>shows processes running in the current terminal</li> </ul> <strong>ps aux</strong> <ul> <li>shows all running processes</li> </ul> <strong>kill process_id</strong> <ul> <li>stops a process using its process ID</li> </ul> <strong>kill -9 process_id</strong> <ul> <li>forcefully stops a process</li> </ul> <strong>jobs</strong> <ul> <li>shows programs running in the background from the current terminal</li> </ul> <strong>command &amp;</strong> <ul> <li>runs a command in the background</li> </ul> <strong>Ctrl + C</strong> <ul> <li>stops the command currently running in the terminal</li> </ul> <strong>Ctrl + Z</strong> <ul> <li>pauses the command currently running</li> </ul> <strong>fg</strong> <ul> <li>returns a paused or background process to the foreground</li> </ul> </div>

<div style="background-color: #FFDFBE; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Network Commands</h3> 
<br>    
<strong>ping google.com</strong> <ul> <li>tests whether your computer can connect to another device or website</li> </ul> <strong>ip addr</strong> <ul> <li>displays network interfaces and IP addresses</li> </ul> <strong>ip link</strong> <ul> <li>displays network interface information</li> </ul> <strong>ifconfig</strong> <ul> <li>displays network information on older systems if installed</li> </ul> <strong>hostname -I</strong> <ul> <li>displays the device's IP address</li> </ul> <strong>curl website_url</strong> <ul> <li>retrieves information from a website or server</li> </ul> <strong>wget website_url</strong> <ul> <li>downloads a file from a website</li> </ul> <strong>ssh username@ip_address</strong> <ul> <li>remotely connects to another Linux computer or development board</li> </ul> <strong>scp file.txt username@ip_address:/destination/path</strong> <ul> <li>securely copies a file to another computer</li> </ul> <strong>scp username@ip_address:/path/file.txt .</strong> <ul> <li>securely copies a file from another computer into the current directory</li> </ul> </div>

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; ">
<h3 style="font-size: 20px; font-weight: bold;">Installing Software</h3> 
<br>    
<strong>sudo apt update</strong> <ul> <li>updates the list of available software packages on Ubuntu or Debian systems</li> </ul> <strong>sudo apt upgrade</strong> <ul> <li>installs available updates for installed packages</li> </ul> <strong>sudo apt install package_name</strong> <ul> <li>installs a software package</li> </ul> <strong>sudo apt remove package_name</strong> <ul> <li>removes a software package</li> </ul> <strong>apt list --installed</strong> <ul> <li>lists installed packages</li> </ul> <strong>pip install package_name</strong> <ul> <li>installs a Python package</li> </ul> <strong>pip3 install package_name</strong> <ul> <li>installs a Python package for Python 3</li> </ul> </div>

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-family: 'Times New Roman', Times, serif; font-size: 16px; "> 
<h3 style="font-size: 20px; font-weight: bold;">Archives and Compression</h3> 
<br>    
<strong>tar -cvf archive.tar folder_name</strong> <ul> <li>combines a folder into a tar archive</li> </ul> <strong>tar -xvf archive.tar</strong> <ul> <li>extracts a tar archive</li> </ul> <strong>tar -czvf archive.tar.gz folder_name</strong> <ul> <li>compresses a folder into a tar.gz archive</li> </ul> <strong>tar -xzvf archive.tar.gz</strong> <ul> <li>extracts a tar.gz archive</li> </ul> <strong>zip -r archive.zip folder_name</strong> <ul> <li>compresses a folder into a ZIP file</li> </ul> <strong>unzip archive.zip</strong> <ul> <li>extracts a ZIP file</li> </ul> </div>

<h3 style="font-size: 20px; font-weight: bold;">Git Commands</h3> 
<br>    
<strong>git status</strong> <ul> <li>shows changed, staged, and untracked files</li> </ul> <strong>git init</strong> <ul> <li>creates a new Git repository in the current folder</li> </ul> <strong>git clone repository_url</strong> <ul> <li>downloads a GitHub repository</li> </ul> <strong>git add file_name</strong> <ul> <li>stages one file for a commit</li> </ul> <strong>git add .</strong> <ul> <li>stages all changed files</li> </ul> <strong>git commit -m "message"</strong> <ul> <li>saves staged changes with a commit message</li> </ul> <strong>git push</strong> <ul> <li>uploads local commits to GitHub</li> </ul> <strong>git pull</strong> <ul> <li>downloads and combines changes from GitHub</li> </ul> <strong>git log</strong> <ul> <li>displays previous commits</li> </ul> <strong>git log --oneline</strong> <ul> <li>displays commits in a shorter format</li> </ul> <strong>git branch</strong> <ul> <li>lists branches</li> </ul> <strong>git branch branch_name</strong> <ul> <li>creates a new branch</li> </ul> <strong>git switch branch_name</strong> <ul> <li>switches to another branch</li> </ul> </div>
