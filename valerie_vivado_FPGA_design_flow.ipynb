<div style="color: hotpink; font-size: 30px; font-family: Times; font-weight: bold;">
Vivado FPGA Design Flow</div>

<div style="background-color: #f3e5ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
This notebook will explain the Vivado FPGA design flow.</div>

<div style="background-color: #c1eeed; padding: 12px; border-radius: 10px; font-size: 20px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Overall Flow:</b></h3>

<div style="background-color: #c1eeed; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul>RTL → Synthesis → LUTs → Implementation → FPGA → Bitstream</ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>RTL:</b></h3>
<l>your hardware description </l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>where you describe the circuit you want using HDL such as Verilog/VHDL</li> </ul>
<ul> <li>a description of the circuit, not yet mapped to the actual resources inside the FPGA</li> </ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Synthesis:</b></h3>
<l>converts your RTL code into actual digital logic elements</l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>figures out what hardware is needed to perform what you describe</li> </ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>LUT (lookup table):</b></h3>
<l>one of the basic building blocks inside an FPGA</l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>instead of having separate physical logic gates, an FPGA can program a LUT to behave like different logic functions</li> </ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Implementation:</b></h3>
<l>figures out placement & routing</l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li><b>placement:</b> where should each LUT, flip-flop, etc. be placed inside the FPGA</li> </ul>
    <ul> <li><b>routing:</b> how should these components be connected together</li> </ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Physical FPGA:</b></h3>
<l>using actual resources inside the FPGA chip </l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>after implementation, Vivado will have determined how your design uses the actual resources inside the FPGA chip</li> </ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Simulation:</b></h3>
<l>testing your design</l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>simulation lets you check:</li> </ul>
    <ul> <ul><ul><li>inputs </li> </ul></ul></ul>
    <ul> <ul><ul><li> RTL design</li> </ul></ul></ul>
    <ul> <ul><ul><li>exepcted outputs </li> </ul></ul></ul>
    <l><b>without needing to physically program the FPGA first</b></l>

<br><ul> <li>overlay: PYNQ overlay is the programmable hardware design that you load onto the FPGA</li> </ul>
    <ul> <ul><ul><li> .bit bitstream is a key part of that overlay</li> </ul></ul></ul>
    <ul> <ul><ul><li> you're essentially telling PYNQ to load this hardware configuration onto the FPGA</li> </ul></ul></ul>

</div>

<div style="background-color: #e0f7f6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Bitstream:</b></h3>
<l>the important file at the end of the Vivado flow</l>
    
<div style="background-color: #effbfa; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>bitstream contains the configuration information that tells the FPGA things like:</li> </ul>
    <ul> <ul><ul><li>configure this LUT to perform this function </li> </ul></ul></ul>
    <ul> <ul><ul><li> connect these resources together</li> </ul></ul></ul>
    <ul> <ul><ul><li>use these FPGA pins </li> </ul></ul></ul>
    <ul> <ul><ul><li> configure these registers/resources</li> </ul></ul></ul>

<br><ul> <li>overlay: PYNQ overlay is the programmable hardware design that you load onto the FPGA</li> </ul>
    <ul> <ul><ul><li> .bit bitstream is a key part of that overlay</li> </ul></ul></ul>
    <ul> <ul><ul><li> you're essentially telling PYNQ to load this hardware configuration onto the FPGA</li> </ul></ul></ul>

</div>
