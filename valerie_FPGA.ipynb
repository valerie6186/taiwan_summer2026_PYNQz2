<div style="background-color: #e2d6c4; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Field Programmable Gate Array (FPGA) Structure:</h3>
   
<ul> <li>an integrated circuit that can be configured by the user to create custom digital hardware</li> </ul>
<ul> <li>contains programmable hardware resources that can be connected together to build a circuit </li> </ul>

</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Configurable Logic Blocks (CLB):</h3>
<ul> <li>the main areas of an FPGA where digital logic and data storage are created</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <b>CLBs can be used to create:</b>
<br><ul> <li>logic gates</li> </ul>
<ul> <li>adders</li> </ul>
<ul> <li>comparators</li> </ul>
<ul> <li>counters</li> </ul>
<ul> <li>registers</li> </ul>
<ul> <li>multiplexers</li> </ul>
<ul> <li>state machines</li> </ul>

<br>
<b>an FPGA contains many CLBs arranged in a grid:</b>

CLB   CLB   CLB   CLB

CLB   CLB   CLB   CLB

CLB   CLB   CLB   CLB
    
<ul> <li>programmable routing connects the CLBs together</li> </ul>

</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Look-Up Tables (LUT):</h3>
<ul> <li>performs combinational logic by storing the outputs of a Boolean truth table</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>LUT can implement logic functions such as:</b>
<br><ul> <li>AND</li> </ul>
<ul> <li>OR</li> </ul>
<ul> <li>XOR</li> </ul>
<ul> <li>NAND</li> </ul>
<ul> <li>comparators</li> </ul>
<ul> <li>multiplexers</li> </ul>
</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Flip-Flops & Registers:</h3>
    
<ul> <li>store binary data and update their values when a clock edge occurs</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>flip-flop stores one bit of data, multiple flip-flops can be combined to create a register that stores several bits</li> </ul>
<b>Flip-Flops can be used to create:</b>
<ul> <li>registers</li> </ul>
<ul> <li>counters</li> </ul>
<ul> <li>shift registers</li> </ul>
<ul> <li>state machines</li> </ul>
<ul> <li>data pipelines</li> </ul>

<br>
<b>LUTs perform calculations, while flip-flops store the results:</b>

<ul> <li>input → LUT logic → flip-flop → output</li> </ul>


</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Programmable Routing:</h3>
    
<ul> <li>provides the wires and connections that allow FPGA components to communicate with one another</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>Routing can connect:</b>
<ul> <li>LUTs to flip-flops</li> </ul>
<ul> <li>one configurable logic block to another</li> </ul>
<ul> <li>logic blocks to memory</li> </ul>
<ul> <li>logic blocks to DSP blocks</li> </ul>
<ul> <li>logic blocks to input/output pins</li> </ul>

<br>
<b>LUTs perform calculations, while flip-flops store the results:</b>

<ul> <li>placement & routing: during FPGA implementation, the design software decides where each component should be placed and how the components should be connected</li> </ul>
<ul> <li>without programmable routing, the FPGA components would not be able to work together as one complete circuit</li> </ul>


</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Input/Output Blocks (IOB):</h3>
    
<ul> <li>connect the internal FPGA circuit to external devices and physical pins</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>FPGA pins may be configured as:</b>
<ul> <li>inputs</li> </ul>
<ul> <li>outputs</li> </ul>
<ul> <li>bidirectional signals</li> </ul>
<b>they can connect the FPGA to:</b>
<ul> <li>pushbuttons</li> </ul>
<ul> <li>switches</li> </ul>
<ul> <li>LEDs</li> </ul>
<ul> <li>sensors</li> </ul>
<ul> <li>displays</li> </ul>
<ul> <li>motors</li> </ul>
<ul> <li>communication devices</li> </ul>


</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Block RAM (BRAM):</h3>
    
<ul> <li>dedicated memory inside the FPGA that efficiently stores larger amounts of data</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>BRAM can store:</b>
<ul> <li>program data</li> </ul>
<ul> <li>instructions</li> </ul>
<ul> <li>images</li> </ul>
<ul> <li>audio samples</li> </ul>
<ul> <li>sensor readings</li> </ul>
<ul> <li>communication buffers</li> </ul>
<ul> <li>look-up tables</li> </ul>
<b>BRAM can be configured with different:</b>
<ul> <li>memory sizes</li> </ul>
<ul> <li>data widths</li> </ul>
<ul> <li>memory depths</li> </ul>
<ul> <li>read & write ports</li> </ul>


</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Digital Signal Processing (DSP):</h3>
    
<ul> <li>specialized hardware components that perform mathematical operations quickly and efficiently</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>DSP blocks can perform operations such as:</b>
<ul> <li>multiplication</li> </ul>
<ul> <li>additions</li> </ul>
<ul> <li>subtraction</li> </ul>
<ul> <li>multiply accumulate operations</li> </ul>
<ul> <li>signal filtering</li> </ul>
<b>they are commonly used in:</b>
<ul> <li>audio processing</li> </ul>
<ul> <li>image processing</li> </ul>
<ul> <li>robotics</li> </ul>
<ul> <li>motor control</li> </ul>
<ul> <li>machine learning</li> </ul>
<ul> <li>communication systems</li> </ul>

</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Hard Processing Blocks:</h3>
    
<ul> <li>permanent, specialized components already built into some FPGA devices</li> </ul>
<ul> <li>unlike LUTs and flip-flops, hard blocks are not created from programmable logic</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>Hard Processing Block examples:</b>
<ul> <li>ARM processors</li> </ul>
<ul> <li>memory controllers</li> </ul>
<ul> <li>ethernet controllers</li> </ul>
<ul> <li>PCI express controllers</li> </ul>
<ul> <li>high-speed communication transceivers</li> </ul>

</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Configuration Memory & Bitstream:</h3>
    
<ul> <li>stores the information that tells the FPGA how its logic and connections should be programmed</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>FPGA configuration determines:</b>
<ul> <li>what each LUT does</li> </ul>
<ul> <li>how the routing is connected</li> </ul>
<ul> <li>how registers behave</li> </ul>
<ul> <li>how input/output pins operate</li> </ul>
<ul> <li>how memory and DSP blocks are configured</li> </ul>

    
<br> 
<ul><li><b>bitstream:</b> the file used to program the FPGA</li></ul>
<ul>HDL design → synthesis → implementation → bitstream</ul>
<ul><li>when the bitstream is loaded, the FPGA becomes the digital circuit described by the HDL code</li></ul>
<ul><li>many FPGAs lose their configuration when power is removed, so the bitstream must be loaded again when the board starts</li></ul>


</div>

<div style="background-color: #e9e0d2; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">How HDL Uses FPGA Resources:</h3>
    
<ul> <li>when HDL code is synthesized, the FPGA design software converts the code into available hardware resources</li> </ul>
    
<div style="background-color: #f1eae1; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>boolean expression → LUT</li> </ul>
<ul> <li>clocked assignment → flip-flop</li> </ul>
<ul> <li>counter → LUT & flip-flops</li> </ul>
<ul> <li>large memory → block RAM</li> </ul>
<ul> <li>multiplication → DSP block</li> </ul>
<ul> <li>external signal → input/output block</li> </ul>
<ul> <li>signal connections → programmable routing</li> </ul>

    
<br> 
<ul><li><b>bitstream:</b> the file used to program the FPGA</li></ul>
<ul>HDL design → synthesis → implementation → bitstream</ul>
<ul><li>when the bitstream is loaded, the FPGA becomes the digital circuit described by the HDL code</li></ul>
<ul><li>many FPGAs lose their configuration when power is removed, so the bitstream must be loaded again when the board starts</li></ul>


</div>

<div style="background-color: #9bd0ff; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
FPGA vs PLD:</div>

<div style="background-color: #7ed4e6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Programmable Logic Device (PLD):</h3>
   
<ul> <li>digital device that can be programmed to perform a specific logic function</li> </ul>
<ul> <li>allows a designer to create custom hardware without manufacturing a completely new integrated circuit</li> </ul>

</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Relationship Between FPGAs & PLDs:</h3>
    
<ul> <li>PLD refers to the entire family of programmable logic devices, while FPGA refers to one specific and more advanced type</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>an FPGA is a type of PLD</li> </ul>
<ul> <li>every FPGA is a PLD, but not every PLD is an FPGA</li> </ul>

</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Simple PLDs:</h3>
    
<ul> <li>small programmable devices used for basic Boolean logic and control functions</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>simple PLD examples:</b>
<ul> <li>PAL</li> </ul>
<ul> <li>PLA</li> </ul>
<ul> <li>GAL</li> </ul>

    
<br><b>they are commonly used to create:</b>
<ul> <li>logic gates</li> </ul>
<ul> <li>small decoders</li> </ul>
<ul> <li>address-selection circuits</li> </ul>
<ul> <li>simple control logic</li> </ul>
<ul> <li>small state machines</li> </ul>

<br><ul> <li>simple PLDs contain fewer logic resources than FPGAs </li> </ul>
<ul> <li>they are suitable when the design only requires a small number of inputs, outputs, and logic equations</li> </ul>

</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Simple PLDs:</h3>
    
<ul> <li>small programmable devices used for basic Boolean logic and control functions</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>simple PLD examples:</b>
<ul> <li>PAL</li> </ul>
<ul> <li>PLA</li> </ul>
<ul> <li>GAL</li> </ul>

    
<br><b>they are commonly used to create:</b>
<ul> <li>logic gates</li> </ul>
<ul> <li>small decoders</li> </ul>
<ul> <li>address-selection circuits</li> </ul>
<ul> <li>simple control logic</li> </ul>
<ul> <li>small state machines</li> </ul>

<br><ul> <li>simple PLDs contain fewer logic resources than FPGAs </li> </ul>
<ul> <li>they are suitable when the design only requires a small number of inputs, outputs, and logic equations</li> </ul>

</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Complex PLDs (CPLD):</h3>
    
<ul> <li>a medium-sized programmable device built from several logic blocks connected through programmable routing</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<b>they are commonly used for:</b>
<ul> <li>startup control</li> </ul>
<ul> <li>reset circuits</li> </ul>
<ul> <li>address decoding</li> </ul>
<ul> <li>interface control</li> </ul>
<ul> <li>small state machines</li> </ul>
<ul> <li>combining several simple logic functions</li> </ul>

<br><ul> <li>CPLDs commonly use larger logic structures called macrocells</li> </ul>
<ul> <li><b>macrocells</b>: a programmable logic block in a PLD or CPLD that performs a logic function and can optionally store the result</li> </ul>    
    
</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Logic Capacity:</h3>
    
<ul> <li>how much digital hardware can be created inside the device</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>a basic PLD may only contain enough resources for a small circuit</li> </ul>
<ul> <li>an FPGA may contain thousands or millions of programmable logic elements</li> </ul>
    
</div>

<div style="background-color: #b9deff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Routing:</h3>
    
<ul> <li>determines how the programmable logic components are connected</li> </ul>
    
<div style="background-color: #d1e9ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>PLDs and CPLDs usually have simpler routing structures, making their timing more predictable</li> </ul> 
<ul> <li>FPGAs contain much larger routing networks because they must connect many logic blocks, memories, DSP units, and input/output pins, providing more flexibility, but also making placement and timing more complex</li> </ul>
    
</div>
