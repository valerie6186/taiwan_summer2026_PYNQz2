<div style="background-color: #ffdfbf; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>ROS (Robot Operating System):</b></h3>
<l>software framework used to build and control robotic systems</l>
    
<div style="background-color: #ffecda; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li><b>it provides:</b> </li> </ul>
    <ul> <ul><ul><li> tools</li> </ul></ul></ul>
    <ul> <ul><ul><li> libraries</li> </ul></ul></ul>
    <ul> <ul><ul><li> communication methods</li> </ul></ul></ul>
    <ul> <ul><ul><li> software structure</li> </ul></ul></ul>
    
<l><ul><li>these allow different parts of a robot to communicate & work together</li> </ul></l>
<ul><li>ROS is commonly used with Linux & Ubuntu</li> </ul>


<br><ul> <li>a robot may contain several different components at the same time, without ROS, the programmer would need to manually create a way for all of these components to communicate</li> </ul>
    
</div>

<div style="background-color: #fff1d4; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Nodes:</b></h3>
<l>an individual program running inside a ROS system</l>
    
<div style="background-color: #fff7e5; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>instead of writing one very large program to control the entire robot, ROS divides the robot into many smaller programs called nodes</li> </ul>
<ul> <li>each node usually performs one specific task: </li> </ul>
    <ul> <ul><ul><li> camera node → reads images</li> </ul></ul></ul>
    <ul> <ul><ul><li> IMU node → reads motion data</li> </ul></ul></ul>
    <ul> <ul><ul><li> LiDAR node → measures distances</li> </ul></ul></ul>
    <ul> <ul><ul><li> navigation node → determines where the robot should move</li> </ul></ul></ul>
    <ul> <ul><ul><li> motor node → controls the motors</li> </ul></ul></ul>

<l><ul><li>a robot can have multiple nodes running at the same time</li> </ul></l>

    
</div>

<div style="background-color: #fff1d4; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Topics:</b></h3>
<l>a communication channel used by ROS nodes to exchange data</l>
    
<div style="background-color: #fff7e5; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>nodes can either: </li> </ul>
    <ul> <ul><ul><li> publish data to a topic</li> </ul></ul></ul>
    <ul> <ul><ul><li> subscribe to data from a topic</li> </ul></ul></ul>

<l><ul><li>this is called the <b>publisher-subscriber model</b> </li> </ul></l>
    
<br><ul><li><b>publisher:</b> node that sends data to a ROS topic</li> </ul>
<ul><li><b>subscriber:</b> node that receives data from a ROS topic</li> </ul>
    
</div>

<div style="background-color: #fff1d4; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Services:</b></h3>
<l>used when one node sends a request to another node and waits for a response</l>
    
<div style="background-color: #fff7e5; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>communication follows: request → response </li> </ul>
<ul><li>services are usually used for tasks that happen occasionally rather than continuously</li> </ul>
    
    
</div>

<div style="background-color: #fff1d4; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Actions:</b></h3>
<l>used for tasks that take longer to complete</l>
    
<div style="background-color: #fff7e5; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>actions allow a node to: </li> </ul>
    <ul> <ul><ul><li> send a goal</li> </ul></ul></ul>
    <ul> <ul><ul><li> receive progress updates</li> </ul></ul></ul>
    <ul> <ul><ul><li> receive the final result</li> </ul></ul></ul>
    <ul> <ul><ul><li> cancel the task</li> </ul></ul></ul>


</div>

<div style="background-color: #ffdfbf; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>ROS Packages:</b></h3>
<l>a collection of related ROS software</l>
    
<div style="background-color: #ffecda; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>a package can contain: </li> </ul>
    <ul> <ul><ul><li> nodes</li> </ul></ul></ul>
    <ul> <ul><ul><li> python programs</li> </ul></ul></ul>
    <ul> <ul><ul><li> C++ programs</li> </ul></ul></ul>
    <ul> <ul><ul><li> configuration files</li> </ul></ul></ul>
    <ul> <ul><ul><li> message definitions</li> </ul></ul></ul>
    <ul> <ul><ul><li> launch files</li> </ul></ul></ul>
    <ul> <ul><ul><li> robot models</li> </ul></ul></ul>

<l><ul><li>packages help keep large robotic projects organized</li> </ul></l>

</div>

<div style="background-color: #ffdfbf; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>ROS Workspace:</b></h3>
<l>the main environment where ROS packages are stored and built</l>
    
<div style="background-color: #ffecda; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<ul> <li>a ROS 2 workspace may look like: </li> </ul>
 <ul><ul><ul>ros2_ws/ </ul></ul></ul>
 <ul><ul><ul><l>│</l> </ul></ul></ul>
 <ul><ul><ul><l>├── src/</l> </ul></ul></ul>
 <ul><ul><ul><ul><l>│   ├── package_1/</l> </ul></ul></ul></ul>
 <ul><ul><ul><ul><l>│   ├── package_2/</l> </ul></ul></ul></ul>
 <ul><ul><ul><ul><l>│   └── package_3/</l> </ul></ul></ul>
    
 <ul><ul><l>│
<ul>├── build/</ul>
<ul>├── install/</ul>
<ul>└── log/</ul>
   
</div>

<div style="background-color: #ffdfbf; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>ROS 1 vs ROS 2:</b></h3>
<l>two major versions of ROS</l>
    
<div style="background-color: #ffecda; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li><b>ROS 1:</b> original ROS system</li> </ul>
    <ul> <li><b>ROS 2:</b> newer generation of ROS</li> </ul>

<br><ul> <li>ROS 2 is designed to improve: </li> </ul>
    <ul> <ul><ul><li> reliability</li> </ul></ul></ul>
    <ul> <ul><ul><li> security</li> </ul></ul></ul>
    <ul> <ul><ul><li> real-time capabilities</li> </ul></ul></ul>
    <ul> <ul><ul><li> distributed systems</li> </ul></ul></ul>
    <ul> <ul><ul><li> multi-robot systems</li> </ul></ul></ul>
    <ul> <ul><ul><li> embedded system integration</li> </ul></ul></ul>

</div>
