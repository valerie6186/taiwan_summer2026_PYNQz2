ROS Multi PYNQ Communication</div>

<div style="background-color: #f3e5ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
This notebook will demonstrate how to establish ROS 1 communication between two PYNQ-Z2 boards.</div>

<div style="text-align: center;">
    <img src="images/pynqcomsetup.png" width="400">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 1):</b> connect both PYNQ-Z2 boards to the same network</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on each PYNQ-Z2, open a terminal and enter: <b>hostname -I</b></li> </ul>
    <ul> <ul><ul><li> this will display the boards' IP addresses</li> </ul></ul></ul>
    <ul> <ul><ul><li> write down the IP address of each board because they will be needed later</li> </ul></ul></ul>

   
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 2):</b> choose & configure ROS Master</h3>
   
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>choose one board to be PYNQ Master and enter: </li> </ul> 
    <ul>export ROS_MASTER_URI=http://192.167.0.100:11311</ul>
    <ul>export ROS_IP=192.167.0.100</ul>
   
<ul> <ul><ul><li>ROS_MASTER_URI tells ROS where the ROS Master is</li> </ul></ul></ul>
<ul> <ul><ul><li>ROS_IP tells ROS the IP address of the current board</li> </ul></ul></ul>
<ul> <ul><ul><li>replace 192.167.0.100 with the IP address of your chosen ROS Master</li> </ul></ul></ul>
</div>
   

</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 3):</b> start ROS Master</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on PYNQ Master enter: <b>roscore</b></li> </ul>
    <l>
<ul> <ul><ul><li>this starts the ROS Master and allows ROS nodes on both boards to discover each other</li> </ul></ul></ul>
<ul> <ul><ul><li>leave this terminal running while communicating between the boards</li> </ul></ul></ul>
    
<div style="text-align: center;">
    <img src="images/masterroscore.png" width="500">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 4):</b> configure second PYNQ board</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on second PYNQ enter:</li> </ul>
    <ul>export ROS_MASTER_URI=http://192.167.0.100:11311</ul>
    <ul>echo $ROS_MASTER_URI</ul>
    <ul>export ROS_IP=192.167.0.104</ul>
   
<ul> <ul><ul><li>both boards have the same ROS_MASTER_URI</li> </ul></ul></ul>
<ul> <ul><ul><li>each board has its own ROS_IP</li> </ul></ul></ul>
    
<div style="text-align: center;">
    <img src="images/configureboard2.png" width="700">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 5):</b> verify communication with ROS Master</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on second PYNQ enter: <b>rosnode list</b></li> </ul>
    
   
<ul> <ul><ul><li>if the connection is successful, /rosout should appear</li> </ul></ul></ul>
    
<div style="text-align: center;">
    <img src="images/roslist.png" width="800">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 6):</b> publish message from PYNQ Master</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>open a new terminal on PYNQ Master and configure the ROS environment in the new terminal:</li> </ul>
    <ul> <li>also publish a message to the /pynq_test topic</li> </ul>
    <ul>export ROS_MASTER_URI=http://192.167.0.100:11311</ul>
    <ul>export ROS_IP=192.167.0.100</ul>
    <ul>rostopic pub -r 1 /pynq_test std_msgs/String "data: 'hello from PYNQ 1'"</ul>
 
    
<div style="text-align: center;">
    <img src="images/testmessage1.png" width="500">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 7):</b> check ROS topic on second PYNQ</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on second PYNQ enter: <b>rostopic list</b></li> </ul>
    
   
<ul> <ul><ul><li>/pynq_test should now appear in the list of available ROS topics, confirming that the topic published by the PYNQ Master has been registered with the ROS Master</li> </ul></ul></ul>
    
<div style="text-align: center;">
    <img src="images/rostopic.png" width="800">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 8):</b> receive message on second PYNQ</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>on second PYNQ enter: <b>rostopic echo /pynq_test</b></li> </ul>
    
   
<ul> <ul><ul><li>second PYNQ will subscribe to /pynq_test and display the transmitted data</li> </ul></ul></ul>
<ul> <ul><ul><li>because the publisher is using -r 1, a message should appear approximately once every second</li> </ul></ul></ul>

    
<div style="text-align: center;">
    <img src="images/messagereceived.png" width="800">
</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>ADDITIONAL NOTE:</b> controlling transmission timing</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul><li>-r 1  → 1 message per second</li></ul>
<ul><li>-r 2  → 2 messages per second</li></ul>
<ul><li>-r 5  → 5 messages per second</li></ul>
<ul><li>-r 10 → 10 messages per second</li></ul>
    
   
    

</div>

<div style="background-color: #ecdcfe; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px;"><b>Step 9):</b> stop communication</h3>
    
<div style="background-color: #f9f3ff; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
    <ul> <li>press Ctrl + C in the publishing terminal to stop transmitting messages</li> </ul>
    <ul> <li>press Ctrl + C in the rostopic echo terminal to stop displaying received messages
</li> </ul>
    <ul> <li>press Ctrl + C in the ROS Master terminal when completely finished</li> </ul>
 
 
</div>
