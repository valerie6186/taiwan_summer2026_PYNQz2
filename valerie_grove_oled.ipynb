<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
3) Understanding OLED Programming
</div> 

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Adapter Configuration</h3>
<ul> <li>this creates the connection between the PYNQ board and OLED</li> </ul>
</div>

```python
adapter = ArduinoSEEEDGroveAdapter(base.ARDUINO, I2C='grove_oled')
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Define Device Object
</div>

```python
oled = adapter.I2C
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Initialize the Grove I2C OLED</h3>
<ul> <li>sets up OLED using its normal default settings</li> </ul>
</div>

```python
oled.set_default_config()
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Normal Display</h3>
<ul> <li>this places OLED in normal display mode: dark background, light text</li> </ul>
</div>

```python
oled.set_normal_display()
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Inverse Display</h3>
<ul> <li>this reverses OLED's normal display mode: light background, dark text</li> </ul>
</div>

```python
oled.set_inverse_display()
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Display a Character on OLED</h3>
<ul> <li>this displays one character using its hexadecimal ASCII value</li> </ul>
<ul> <li>0x means the number is written in hexadecimal, hexadecimal 56 equals decimal 86, ASCII character 86 is uppercase V</li> </ul>
</div>

```python
oled.put_char(0x56)
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Display a String on OLED</h3>
<ul> <li>this displays an entire string</li> </ul>
</div>

```python
oled.put_string("Hello World !")
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;">Clear OLED Display</h3>
<ul> <li>this removes the currently displayed text and pixels</li> </ul>
<ul> <li>without clearing the display, new text might overlap or appear after the previous text</li> </ul>
<ul> <li>clearing the screen does not always reset mode settings, those may need to changed separately</li> </ul>
</div>

```python
oled.clear_display()
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Display Position</h3>
<ul> <li>two numbers select where writing should begin on the display</li> </ul>
<ul> <li>typical starting position is at (0,0)</li> </ul>
<ul> <li>first number selects the display row position</li> </ul>
<ul> <li>second number selects the display column position</li> </ul>
</div>


```python
oled.set_position(3, 15)
oled.put_string("Hello World !")
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Gray Level</h3>
<ul> <li>sets the pixel brightness level</li> </ul>
<ul> <li>range: 0x00 to 0xFF, lowest to highest</li> </ul>
</div>


```python
oled.clear_display()
oled.set_gray_level(0xFF)
oled.put_string("Hello World !")
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Page Mode (same row wrapping)</h3>
<ul> <li>writing wraps around on the same row instead of automatically moving down to the next row</li> </ul>
</div>


```python
oled.clear_display() 
oled.set_same_row_wrap_mode() 
oled.put_string("Hello World !, Have a good day") 
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Horizontal Mode (next row wrapping)</h3>
<ul> <li>writing continues on the next row when it reaches the end of the current row</li> </ul>
</div>

```python
oled.clear_display()
oled.set_next_row_wrap_mode() 
oled.put_string("Hello World !, Have a good day") 
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Setting Contrast Level</h3>
<ul> <li>sets OLED’s overall contrast level</li> </ul>
<ul> <li>range: 0x00 to 0xFF, lowest to highest</li> </ul>
<ul> <li>avoid leaving OLED unnecessarily bright for long periods</li> </ul>
</div>

```python
oled.clear_display()
oled.set_contrast_level(0xFF)
oled.put_string("Hello World !")
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Activate Scroll</h3>
<ul> <li>initiates OLED’s scrolling behavior</li> </ul>
</div>

```python
oled.clear_display()
oled.activate_scroll()
oled.put_string("Hello World !")
```

<div style="background-color: #B0E0E6; padding: 12px; border-radius: 10px; font-size: 16px; font-family: Times;">
<h3 style="font-size: 20px; font-weight: bold;"> Deactivate Scroll</h3>
<ul> <li>terminates OLED’s scrolling behavior</li> </ul>
</div>

```python
oled.clear_display()
oled.deactivate_scroll()
oled.put_string("Hello World !")
```

<div style="background-color: #ffe6f2; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
4) Building an Interactive OLED Interface
</div> 

<div style="background-color: #C1E1C1; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Step 1: Reset/Configure OLED
</div> 

```python
adapter = ArduinoSEEEDGroveAdapter(base.ARDUINO, I2C='grove_oled')
oled = adapter.I2C

# stop any scrolling from the previous examples
oled.deactivate_scroll()

# restore the default oled configuration
oled.set_default_config()

# return to the normal display mode
oled.set_normal_display()

# allow long text to continue onto the next row
oled.set_next_row_wrap_mode()

# restore normal brightness settings
oled.set_gray_level(0xFF)
oled.set_contrast_level(0x80)

# clear old text and return the cursor to the beginning
oled.clear_display()
oled.set_position(0, 0)
```

<div style="background-color: #C1E1C1; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Step 2: Ask User for Name
</div> 

```python
name = input("Enter your name: ")
```

<div style="background-color: #C1E1C1; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Step 3: Check User Input
</div> 

```python
name = name.strip()

if name == "":
    name = "Guest"
```

<div style="background-color: #C1E1C1; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Step 4: Create Greeting
</div> 

```python
message = "Hello, " + name + "!"
```

<div style="background-color: #C1E1C1; padding: 12px; border-radius: 10px; font-size: 20px; font-weight: bold; font-family: Times;">
Step 5: Display Message
</div> 

```python
oled.put_string(message)
print("Displayed on OLED:", message)
```
